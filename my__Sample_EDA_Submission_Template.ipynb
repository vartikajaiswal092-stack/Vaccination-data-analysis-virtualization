## ***4. Data Vizualization, Storytelling & Experimenting with charts : Understand the relationships between variables***

#### Chart - 1

````python
# Chart - 1 visualization code

coverage_column = [col for col in coverage_data.columns
                   if str(col).strip().lower() == "coverage"][0]

plt.figure(figsize=(10, 6))

plt.hist(coverage_data[coverage_column], bins=30)

plt.title("Distribution of Vaccination Coverage")
plt.xlabel("Vaccination Coverage (%)")
plt.ylabel("Number of Records")

plt.show()
````

##### 1. Why did you pick the specific chart?

**A histogram was selected because it is suitable for understanding the distribution of vaccination coverage values. It helps identify the most common coverage ranges and shows whether the values are concentrated at low, medium, or high coverage levels. text**

##### 2. What is/are the insight(s) found from the chart?

**The chart helps identify the overall distribution of vaccination coverage in the dataset. It shows whether most records have low, moderate, or high vaccination coverage and helps identify unusual coverage values that may require further investigation.**

##### 3. Will the gained insights help creating a positive business impact?
Are there any insights that lead to negative growth? Justify with specific reason.

**Yes, these insights can support positive public health impact by helping health authorities identify areas where vaccination coverage is low and where additional vaccination efforts may be required. Therefore, identifying such gaps can help in better resource allocation and targeted vaccination strategies.**

#### Chart - 2

````python
# Chart - 2 visualization code
year_column = [col for col in coverage_data.columns
               if str(col).strip().lower() == "year"][0]

coverage_column = [col for col in coverage_data.columns
                   if str(col).strip().lower() == "coverage"][0]

yearly_coverage = coverage_data.groupby(year_column)[coverage_column].mean()

plt.figure(figsize=(10, 6))

plt.plot(yearly_coverage.index, yearly_coverage.values, marker="o")

plt.title("Average Vaccination Coverage Over the Years")
plt.xlabel("Year")
plt.ylabel("Average Vaccination Coverage (%)")

plt.xticks(rotation=45)
plt.grid(True)

plt.show()
````

##### 1. Why did you pick the specific chart?

**A line chart was selected because it is suitable for showing changes and trends over time. It helps understand how the average vaccination coverage has changed across different years.**

##### 2. What is/are the insight(s) found from the chart?

**The chart helps identify the overall trend in average vaccination coverage over the years. It can show periods where vaccination coverage increased or decreased and helps identify years that may require further investigation.**

##### 3. Will the gained insights help creating a positive business impact?
Are there any insights that lead to negative growth? Justify with specific reason.

**Yes, these insights can support better public health planning by identifying periods of lower vaccination coverage. A decline in coverage may indicate gaps in vaccination programs and can help health authorities plan targeted interventions and allocate resources more effectively.**

#### Chart - 3

````python
# Chart - 3 visualization code
antigen_column = [col for col in coverage_data.columns
                  if str(col).strip().lower() == "antigen"][0]

coverage_column = [col for col in coverage_data.columns
                   if str(col).strip().lower() == "coverage"][0]

antigen_coverage = coverage_data.groupby(antigen_column)[coverage_column].mean()

antigen_coverage = antigen_coverage.sort_values(ascending=False).head(10)

plt.figure(figsize=(12, 6))

plt.bar(antigen_coverage.index, antigen_coverage.values)

plt.title("Top 10 Antigens by Average Vaccination Coverage")
plt.xlabel("Antigen")
plt.ylabel("Average Vaccination Coverage (%)")

plt.xticks(rotation=45)
plt.show()
````

##### 1. Why did you pick the specific chart?

#### Chart - 14 - Correlation Heatmap

````python
# Correlation Heatmap visualization code
numeric_data = coverage_data.iloc[:, [3, 8, 9, 10]]

correlation = numeric_data.corr()

plt.figure(figsize=(8, 6))

plt.imshow(correlation, cmap="coolwarm", interpolation="nearest")

plt.colorbar()

plt.xticks(range(len(correlation.columns)), correlation.columns, rotation=45)
plt.yticks(range(len(correlation.columns)), correlation.columns)

plt.title("Correlation Heatmap")

plt.show()
````
